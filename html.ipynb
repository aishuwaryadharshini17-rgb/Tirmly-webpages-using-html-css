```python
<input type="text" id="search" placeholder="Search here..." onkeyup="showSuggestions()">
<ul id="result"></ul>

<script>
let items = ["dress", "shoe", "shirt", "shopping bag", "shorts"];

function showSuggestions(){
  let input = document.getElementById("search").value.toLowerCase();
  let result = document.getElementById("result");
  result.innerHTML = "";

  let filtered = items.filter(item => item.includes(input));

  filtered.forEach(item => {
    let li = document.createElement("li");
    li.innerText = item;
    result.appendChild(li);
  });
}
</script>
```
